 # spaCy: Text processing and Pattern matching
 - spaCy is the leading library for NLP : https://spacy.io/
 - spaCy relies on models that are language-specific and come in different sizes. 

```python
# We can load a spaCy model with 'spacy.load'
# https://spacy.io/models/en : 
# English pipeline model optimized::: 'en_core_web_sm'
import spacy
nlp = spacy.load('en_core_web_sm')

# with model loaded process text
doc = nlp("All our dreams can come true, if we have the courage to pursue our dreams")
```

## Tokenizing
 - A token is a unit of text in the document, such as individual words and punctuation.
 - returns a document object that contains tokens.   
 - Iterating through a document gives us token objects
 - Each of these tokens comes with additional information Ex: **token.lemma_ and token.is_stop**

```python
for token in doc:
    print(token)
```

## Text preprocessing
 - **"Lemmatizing"** The "lemma" of a word is its base form. For example, *"walk"* is the lemma of the word *"walking"*. So, when you lemmatize the word walking, you would convert it to walk. Lemmatizing helps by combining multiple forms of the same word into one base form ("calming", "calms", "calmed" would all change to "calm")
 
 - **Stopwords** are words that occur frequently in the language and don't contain much information. English stopwords include "the", "is", "and", "but", "not". . Removing stop words might help the predictive model focus on relevant words.( dreams courge etc.)
 
 - *token.lemma_ returns the lemma*, while *token.is_stop returns a boolean True if the token is a stopword* (and False otherwise).
 
 - -PRON- is the default lemma for pronouns in spaCy (see the docs): About spaCy's custom pronoun lemma. Unlike verbs and common nouns, there's no clear base form of a personal pronoun. 


```python
print(f"Token \t\tLemma \t\tStopword".format('Token', 'Lemma', 'Stopword'))
print("-"*40)
for token in doc:
    print(f"{str(token)}\t\t{token.lemma_}\t\t{token.is_stop}")
```

## Pattern Matching
- To match individual or to match a list of terms we create a 'Matcher', it's easier and more efficient to use 'PhraseMatcher'
- For example, if we want to find where different smartphone models show up in some text, We can create patterns for the model names of interest.

```python
# The matcher is created using the vocabulary of our model. 
# Here we're using the small English model you loaded earlier. Setting attr='LOWER' will match the phrases on lowercased text.
# This provides case insensitive matching.

from spacy.matcher import PhraseMatcher
matcher = PhraseMatcher(nlp.vocab, attr='LOWER')
```

```python
# nlp(text) is created for each term to be found
terms = ['Cadbury', 'Nestle', 'Hershey', 'Amul']
patterns = [nlp(text) for text in terms]
matcher.add("TerminologyList", patterns)
```

```python
# create a document from the text to search and use the phrase matcher to find where the terms occur in the text.
text_doc = nlp("Fascinating review overall, and some really interesting insight.Amul competeting against the Nestle and last year’s market winners of people choice of Cadbury and Hershey.") 
matches = matcher(text_doc)
print(matches)
```

```python
#The matches here are a tuple of the match id and the positions of the start and end of the phrase.
for (match_id,start,end) in matches:
    print(text_doc[start:end])
```
